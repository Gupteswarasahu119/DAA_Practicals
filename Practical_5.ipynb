```python
def knapsack(values, weights, capacity):
    n = len(values)

    # Create DP table
    dp = [[0 for _ in range(capacity + 1)] for _ in range(n + 1)]
        # Build table
    for i in range(1, n + 1):
        for w in range(capacity + 1):
            if weights[i - 1] <= w:
                dp[i][w] = max(
                    values[i - 1] + dp[i - 1][w - weights[i - 1]],
                    dp[i - 1][w]
                )
            else:
                dp[i][w] = dp[i - 1][w]

    return dp[n][capacity]


# Input
n = int(input("Enter the number of items: "))

values = []
weights = []

print("\nEnter value and weight of each item:")

for i in range(n):
    value = int(input(f"Value of item {i + 1}: "))
    weight = int(input(f"Weight of item {i + 1}: "))
    values.append(value)
    weights.append(weight)

capacity = int(input("\nEnter the knapsack capacity: "))

# Output
max_profit = knapsack(values, weights, capacity)

print("\nMaximum Profit =", max_profit)
```

```python
def matrix_chain_order(p):
    n = len(p)

    m = [[0 for _ in range(n)] for _ in range(n)]

    for L in range(2, n):
        for i in range(1, n - L + 1):
            j = i + L - 1

            m[i][j] = float('inf')

            for k in range(i, j):
                q = m[i][k] + m[k + 1][j] + p[i - 1] * p[k] * p[j]

                if q < m[i][j]:
                    m[i][j] = q

    return m[1][n - 1]


n = int(input("Enter number of matrices: "))

p = []

print("Enter dimensions:")

for i in range(n + 1):
    p.append(int(input()))

result = matrix_chain_order(p)

print("Minimum number of multiplications =", result)
```
