```python
def factorial_iterative(n):
    # Handle the error case for negative numbers
    if n < 0:
        return "Factorial is not defined for negative numbers."

    result = 1
    # Loop from 1 to n (inclusive) and multiply
    for i in range(1, n + 1):
        result *= i

    return result

# --- Execution & Verification ---
num = 5
answer_iterative = factorial_iterative(num)
print(f"The factorial of {num} using the Iterative Method is: {answer_iterative}")
```
