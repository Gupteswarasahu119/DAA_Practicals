```python
def floyd_warshall(graph):
    n = len(graph)

    # Create a copy of the graph
    dist = [row[:] for row in graph]

    # Find shortest paths
    for k in range(n):
        for i in range(n):
            for j in range(n):
                if dist[i][k] != float('inf') and dist[k][j] != float('inf'):
                    dist[i][j] = min(
                        dist[i][j],
                        dist[i][k] + dist[k][j]
                    )

    return dist


# Input
n = int(input("Enter number of vertices: "))

print("Enter the adjacency matrix:")
print("Enter 0 for no self-loop and INF for no direct edge.")

graph = []

for i in range(n):
    row = []
    values = input().split()

    for value in values:
        if value.upper() == "INF":
            row.append(float('inf'))
        else:
            row.append(int(value))

    graph.append(row)


# Function Call
result = floyd_warshall(graph)


# Output
print("\nShortest Distance Matrix:")

for row in result:
    for value in row:
        if value == float('inf'):
            print("INF", end="\t")
        else:
            print(value, end="\t")
    print()
```
