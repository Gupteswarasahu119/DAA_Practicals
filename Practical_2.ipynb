```python
# ==========================================
# 1. LINEAR SEARCH IMPLEMENTATION
# ==========================================
def linear_search(arr, target):
    """
    Sequentially checks each element of the array until a match is found.
    Works on both sorted and unsorted arrays.
    """
    for i in range(len(arr)):
        if arr[i] == target:
            return i  # Return the index if found
    return -1  # Return -1 if the target is not in the array


# ==========================================
# 2. BINARY SEARCH IMPLEMENTATION
# ==========================================
def binary_search(arr, target):
    """
    Repeatedly divides a SORTED search interval in half.
    Requires the input array to be sorted.
    """
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        # Check if target is present at mid
        if arr[mid] == target:
            return mid
        # If target is greater, ignore left half
        elif arr[mid] < target:
            low = mid + 1
        # If target is smaller, ignore right half
        else:
            high = mid - 1

    return -1  # Return -1 if the target is not in the array


# ==========================================
# DRIVER CODE & ANSWERS
# ==========================================

# A sorted dataset to test both algorithms
test_array = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
target_value = 23

print(f"Target Element: {target_value}")
print(f"Input Array:    {test_array}\n")

# Execute Linear Search
linear_result = linear_search(test_array, target_value)
print("--- Linear Search Result ---")
if linear_result != -1:
    print(f"Answer: Element found at index {linear_result}")
else:
    print("Answer: Element not found in the array")

# Execute Binary Search
binary_result = binary_search(test_array, target_value)
print("\n--- Binary Search Result ---")
if binary_result != -1:
    print(f"Answer: Element found at index {binary_result}")
else:
    print("Answer: Element not found in the array")
```
